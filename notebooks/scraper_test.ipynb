# Testing Scraper Sources

* paperscraper [https://pypi.org/project/paperscraper/](https://pypi.org/project/paperscraper/) | Python package for web scraping metadata or full-text files from PubMed and servers: arXiv, medRxiv, bioRxiv, and chemRxiv. This will help us gain a deeper understanding beyond what PubMed’s API will give us and give us counts for citations, like in Google Scholar | Free- install the Python package and use it | 30,000 \- 400,000 papers | Depends on how many places we scrape from.
* PRAW library (Python Reddit API Wrapper) for the Reddit API  [https://developers.reddit.com/docs/capabilities/server/reddit-api](https://developers.reddit.com/docs/capabilities/server/reddit-api) | This will help to support retrieving Reddit post data. It can scrape posts, comments, subreddits, users (data we will not use), and searches. | Free for non-commercial research use, though rate-limited | Around 1,000 posts. | Maybe 2 or 3 pulls, depending on the size of the data we want to pull.
* General scraping tools like BeautifulSoup and Scrapy for online forums [https://beautiful-soup-4.readthedocs.io/en/latest/](https://beautiful-soup-4.readthedocs.io/en/latest/) [https://github.com/scrapy/scrapy](https://github.com/scrapy/scrapy)   | Beautiful Soup: Python package, HTML and XML parser.
