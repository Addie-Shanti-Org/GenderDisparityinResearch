```python
# source .venv/bin/activate
from pytrials.client import ClinicalTrials
import pandas as pd
```

```python
ct = ClinicalTrials()
# max_studies limit is 1000
results = ct.get_full_studies(search_expr="endometriosis", max_studies = 50)
```

```python
fields = ct.get_study_fields(
    search_expr="endometriosis",
    fields = ["NCT Number", "Study Title", "Conditions", "Phases", "Interventions", 
              "Primary Outcome Measures", "Enrollment", "Study Status", "Completion Date", "Sponsor"
              ,"Sex", "Age", "Funder Type", "Study Type", "Start Date"],
    max_studies = 1000,
    fmt = "csv"
)
```

```python
endo_df = pd.DataFrame.from_records(fields[1:], columns = fields[0])
print(endo_df)
```

```python
print(endo_df.columns)
```

```python
fields = ct.get_study_fields(
    search_expr='AREA[StartDate]RANGE[1900-01-01, 1900-12-30]',
    fields = ["Start Date", "Completion Date", "NCT Number"],
    max_studies = 1000,
    fmt = "csv"
)

bv_df = pd.DataFrame.from_records(fields[1:], columns = fields[0])
print(bv_df)
```
