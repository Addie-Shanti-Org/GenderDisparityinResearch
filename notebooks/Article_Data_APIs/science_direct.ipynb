```python
response = requests.get(url, headers = headers, params = query_params)
```

```python
if response.status_code == 200:
    data = response.json()
    search_results = data.get('search-results', {}).get('entry', [])
    
    print(search_results)

    for article in search_results:
        title = article.get('dc:title')
        creator = article.get('dc:creator')
        cover_date = article.get('prism:coverDate')
        doi = article.get('prism:doi')
        abstract = article.get('dc:description', 'No Access')

        print(f"Title: {title}")
        print(f"Author: {creator}")
        print(f"Date: {cover_date}")
        print(f"DOI: {doi}")
        print(f"Abstract: {abstract}")

else:
    print(f"Error {response.status_code}")
```
